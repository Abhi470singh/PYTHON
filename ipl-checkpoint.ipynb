## <span id='most-matches' style='font-weight:bold;color:#000080'>5.Players that have played the <b style='color:#FF00FF'> most Matches </b> from 2008-2021 (Top 50) <hr></span>

## <span id='max-avg' style='font-weight:bold;color:#000080'>10.Players with the <b style='color:#FF00FF'> highest Average</b> given that they have played at least 50 Innings(Top 50) <hr></span>

```python
no=df.groupby('Player')['NO'].sum().reset_index()
no.columns=['Player','Number of not outs']

overall_avg=pd.DataFrame()
overall_avg['Player']=runs_s['Player']
overall_avg['Runs']=runs_s['Total Runs']
overall_avg['Inns']=inns['Total Innings Played']
overall_avg['Avg.']=overall_avg['Runs']/(overall_avg['Inns']-no['Number of not outs'])
```

```python
temp=overall_avg.loc[inns['Total Innings Played']>=50]
contenders=temp.shape[0]

temp=temp.sort_values('Avg.',ascending=False)[:50]
plt.figure(figsize=(24,8))
sns.barplot(data=temp,x='Player',y='Avg.',palette='Paired')
plt.xticks(rotation=80);
plt.show()
text="<h3 style='text-align:center'><b style='color:#000080'>"+f"Total Number of Players that satisfied this criterion :</b><b style='color:#FF00FF'>{contenders}</b></h3>"
display(HTML(text))
```
