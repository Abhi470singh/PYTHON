## <span style="color:blue">Population(N):</span>

- A Population include all the elements from a set a data. The Population is the whole set other set of values or individuals, you are interested in. 

- Population characteristic are mean(`μ`)mu, Standard deviation(`σ`)sigma, Proportion(P), Median, Percentiles etc. the Values of a population characteristic is fixed.

Key points:

- Includes every member
- Usually large
- Difficult, costly, or impossible to study completely

📌 Population parameters (true values):

- Mean → μ (mu)
- Standard deviation → σ (sigma)
- Size → N

## <span style="color:yellow"> SAMPLE(n)</span>
- The Sample is subset of the populaction and is the set of values you actually use in your estimation.
- This Sample has some quanity computed from values e.g. mean(`x̄`)x Bar , standared deviation(`s`), sample proportion, etc.  This is Callled sample distributipon.

<b>Examples:</b>

- 1,000 people selected from India
- 200 students from a college
- 500 flat sales from Bangalore

<b>Key points:</b>

- Smaller and manageable
- Saves time and money
- Used to make conclusions about the population

<b>📌 Sample statistics (estimated values):</b>

- Mean → x̄ (x-bar)
- Standard deviation → s
- Size → n

### 3️⃣ Population vs Sample (Comparison)

| Feature         | Population   | Sample            |
| --------------- | ------------ | ----------------- |
| Definition      | Entire group | Part of the group |
| Size            | Very large   | Smaller           |
| Symbol for size | N            | n                 |
| Mean            | μ            | x̄                |
| Std. deviation  | σ            | s                 |
| Accuracy        | Exact        | Approximate       |

### 2️⃣ Median (Middle Value)
- The median is a measure of central tendency that represents the middle value of a dataset when the data is arranged in order.

👉 It tells us where the center of the data lies, especially useful when data has outliers.

<b>🔹 How to find the Median</b>

1️⃣ Arrange the data in <b>ascending order</b>

2️⃣ Two cases

✅ Case 1: Odd number of values

Median = middle value

📌 Example:

Data = 10, 20, 30

Median = 20

✅ Case 2: Even number of values

Median = average of the two middle values

📌 Example:

Data = 10, 20, 30, 40

Median = (20 + 30) / 2 = 25

<b>🔹 Why Median is important</b>
- Not affected by extreme values (outliers)
- Best for skewed data
- Used in income, salary, house price analysis

```python
# case 1 Odd value 
salary = [2000,3000,400,5000,8000,10000,9802]
```

```python
np.sum(salary)
```

```python
len(salary)
```

```python
np.median(salary)
```

```python
# case 2 Even values
sal = [100,102,244,345,345,322]
```

```python
# First the Assending order
l = np.sort(sal)
length = len(sal)
print(l)
length
```

### 3️⃣ Mode (Most Frequent Value)
- Mode is a measure of central tendency that represents the value that occurs most frequently in a dataset.

👉 Simply: jo value sabse zyada baar aaye, wahi mode hoti hai.

🔹 Types of Mode

- Unimodal → one mode
- Bimodal → two modes
- Multimodal → more than two modes
- No mode → all values different

🔹 When mode is useful

- Categorical data (colors, brands)
- Finding most common value
- Market analysis, sales data

1️⃣ Unimodal (one mode)

Only one value appears most often.

2️⃣ Bimodal (two modes)

Two values appear with the same highest frequency.

3️⃣ Multimodal (more than two modes)

More than two values repeat equally.

4️⃣ No Mode

All values occur only once. Nothing stands out.

##### 3️⃣ Standard Deviation
- The Standard deviation is a measure of the ampont of variation or dispersion of a set of values. Alow standard deviation indicates that the values tend to the be close to the mean(also  called the expected values) of the set, while a high Standard deviation indicates that values are spread out over a wide range. 


Population Standard Deviation = $\sqrt{\frac{\sum (x-\mu{})^2}{N}}$


Sample Standard Deviation = $\sqrt{\frac{\sum (x-\bar{x})^2}{n-1}}$

🔍 Symbols meaning

- x → each data value
- μ → population mean
- \bar{x}
→ sample mean
- N → population size
- n → sample size


##### 4️⃣ Variance 
- Variance is a measure of how dataq points differ from the mean. According to Layman, a variable is a measure of how for a set of data(numbers) are spread out from their mean(average) value.

📌 Symbols:

- Population variance → σ²
- Sample variance → s²

Population variance = $\sigma^2 = \frac{\sum (x - \mu)^2}{N}$

Sample variance = $s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$

##### 5️⃣ Interquartile Range (IQR)
-The Interquartile Range (IQR) is a measure of variability that shows the spread of the middle 50% of the data.

👉 It helps us understand how data is spread, while ignoring extreme values (outliers).

🔹 Formula

IQR=Q
3
−Q
1
	​


Where:

Q₁ (First Quartile) → 25% of data

Q₃ (Third Quartile) → 75% of data

🔹 Why IQR is important

- Measures data spread
- Not affected by outliers
- Used in box plots
- Helpful for skewed data

```python
data = [10, 20, 30, 40, 50, 60, 70, 80]

data.sort()

n = len(data)

q1 = (data[n//4 - 1] + data[n//4]) / 2
q3 = (data[3*n//4 - 1] + data[3*n//4]) / 2

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
data = [10, 20, 30, 40, 50, 60, 70, 80]
s = pd.Series(data)

iqr = s.quantile(0.75) - s.quantile(0.25)
print("IQR:", iqr)
```

##### <span style="color:Brown">📊 Percentiles and Quartiles (Statistics)</span>

- Both percentiles and quartiles are measures of position.
They tell us where a value lies in a dataset.

🔹 Quartiles

Quartiles divide the ordered data into 4 equal parts.

Types of Quartiles

| Quartile        | Meaning                             |
| --------------- | ----------------------------------- |
| **Q1**          | 25% of data is below this value     |
| **Q2 (Median)** | 50% of data is below (middle value) |
| **Q3**          | 75% of data is below this value     |

🔹 Percentiles

Percentiles divide the data into 100 equal parts.

Meaning

- P10 → 10% data is below this value
- P50 → 50% data is below (same as median)
- P90 → 90% data is below this value

🔹 Simple Difference Table

| Feature    | Quartiles    | Percentiles          |
| ---------- | ------------ | -------------------- |
| Division   | 4 parts      | 100 parts            |
| Common use | Spread & IQR | Ranking & comparison |
| Examples   | Q1, Q2, Q3   | P10, P50, P90        |

### <span style="color:green">📊 Measurement of Shape</span>
- Measurement of shape tells us data ka distribution kis type ka hai yaani data kaise spread ho raha hai (left, right, symmetric, peaked, flat).

👉 It describes the shape of the data distribution.

🔹 Main Measures of Shape

1️⃣ Skewness

Skewness tells us distribution kis side jhuki hui hai.

🔸 Types of Skewness

- Positive Skewness (Right-skewed)

        Tail right side lambi hoti hai

            👉 Mean > Median > Mode

- Negative Skewness (Left-skewed)

        Tail left side lambi hoti hai

            👉 Mean < Median < Mode

- Zero Skewness (Symmetrical)

        Left aur right same

            👉 Mean = Median = Mode

📌 Example:

Income data → Positive skewed

Exam marks (balanced) → Symmetrical

```python
data = np.random.normal(0,100,100)
```

```python
df = pd.DataFrame({"x":data})
```

```python
df["x"].skew()
```

```python
sns.histplot(x ="x",data=df)
plt.show()
```
