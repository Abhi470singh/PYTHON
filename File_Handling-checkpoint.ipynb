# File Handling in Python

File Handling is a crucial part of any programming language, and Python makes it easy to work with files. Using file handling, yhou can operations loke creating, reading, writing, and appending to files.

## Modes of file Handling

Python provides several modes to work with files:

1. <b>'r' (Read)</b>: Opens the file for reading. if the file does not exist, it throws an error.

2. <b>'w' (Write)</b>: Opens the file for writing. if the file does not exits, it creates a new file. if it exists, it overwrites the content.

3. <b>'a' (Append)</b>: Opens the files for appending new contnt. if the file does not exxist. it creates a new file.

4. <b>'rb'</b>:
Opens the files in binary mode for reading (rb) or writing (wb).

### Using context manager (With)

- It's a good idea to close a file after usage as it will free up the resources
- If we dont close it, garbage collector would close it
- with keyword closes the file as soon as the usage is over

```python
# with
with open('sample2.txt','a') as f:
 f.write('\nSingh Rajput')
f.write('hello')
```

```python
# try f.read() now
with open('sample3.txt','r') as f:
 print(f.read())
```

```python
# seek and tell function
with open ('sample2.txt','r') as f:
 print(f.read(10))
 print(f.read(10))
 print(f.tell())
 f.seek(22)
 print(f.read(10))
 print(f.tell())
```

## Serialization and Deserialization

- <b> Serialization </b>: Process of converting python data types to JSON format

- <b> Deserialization </b> Process of converting JSON to Python data type

- <b> JSON </b> JAVAScript Object Notation (JSON) is a standard text-based format for representing structure data based on JavaScript object syntax

```python
class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname
        self.lname = lname
        self.age = age
        self.gender = gender

    def to_dict(self):
        return {
            "fname": self.fname,
            "lname": self.lname,
            "age": self.age,
            "gender": self.gender
        }

person = Person("Abhishek", "Singh", 23, "Male")

with open("Demo.json", "w") as f:
    json.dump(person.to_dict(), f, indent=4)
```

## Pickling
Pickling is the process whereby a Python object hierarchy is converted into a byte stream, and unpickling is the inverse operation, whereby a byte stream (from a binary
file or bytes-like object) is converted back into an object hierarchy.


```python
class Person:
    
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display_info(self):
        print('Hi my name is',self.name,'and i am ',self.age,'years old')
```

```python
p = Person('Abhishek Singh',20)
```

```python
# pickle dump
import pickle
with open ('person.pkl','wb') as f:
 pickle.dump(p,f)
```

```python
# pickle load
import pickle
with open ('person.pkl','rb') as f:
 p = pickle.load(f)

p.display_info()
```

```python
# Write code here
def get_final_line(filename):
    final_line = ''
    for current_line in open(filename, 'r'):
        final_line = current_line
    return final_line

get_final_line("sample.txt")
```

<b> Q-2: Read thought a text file, line by line.
Use a dict to keep track of how many times each vowel(a,e,i,o and u) appears in the file. Print the resulting tabulation-- dictionary.

```python
# Write code here
def vowel_count_table(filename):
    vowels = ['a', 'e', 'i', 'o', 'u']
    d={i:0 for i in vowels}
    for current_line in open(filename, 'r'):
        for ch in current_line:
            if ch in vowels:
                d[ch] += 1
    return d
vowel_count_table("sample.txt")
```

Python) containing two tab separated columns, with each
column containing a number. Then use Python to read
through the file you’ve created. For each line, multiply each first number by the second and include it in the file in third column. In last add a line Total, by summing the value of third column</b>

input File example: That you need to create

```python
# write code here
file_name = "test-2.txt"
f = open(file_name, "w")
for i in range(1, 11, 2):
    line = '{}\t{}\n'.format(i, i+1)
    f.writelines(line)
f.close()

with open(file_name, 'r') as f:
    lines = f.read().splitlines()
    
total = 0
with open(file_name, 'w') as f:
    for line in lines:
        a,b = line.split(sep="\t")
        res = int(a) * int(b)
        total+= res
        f.write("{}\t{}\t{}\n".format(a,b, res))
    f.write("Total\t"+str(total))
```
