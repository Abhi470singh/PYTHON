```python
def outer_function(text):
    def inner_function():
        # Accessing variable from the outer function's scope
        print(text)
    
    # Returning the inner function
    return inner_function

# Call outer_function and store the returned inner_function
my_function = outer_function("Hello, World!")

# Call the inner function
my_function()  
```

## Decorator Functions with Parameters

Once we understand what decorators in Python are, we can take it a step further by learning how they can be used with function parameters. The main goalof decorators is to enhance or modify the behavior of a function without changing its original code.

Syntax:

func = decorator(func)

Here:

- func is the function we want to decorator
- decorator is the function that wraps and enhances it

Example:

```python
def decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

def say_hello():
    print("Hello!")

# Applying decorator manually
say_hello = decorator(say_hello)

say_hello()
```

### Syntactic Decorator (Using @)
While the above syntax works, it’s not very clear, as it repeats the function name and hides the decorator logic below the function. To fix this, Python introduced a cleaner and more readable syntax using the @ symbol.

Syntax:

@decorator
def func(arg1, arg2, ...):
    pass

Example:

```python
def decorator(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@decorator
def say_hello():
    print("Hello!")

say_hello()
```

### Keeping the Original Name and Docstring of the Decorated Function

In Python, every function has a `__name__` and a `__doc__` attribute, which are used for debugging and documentation purposes. But when we decorate a function, its identity changes to that of the wrapper function, which hides the original metadata.

<b>Example: Without functools.wraps

```python
def decorator(func):
    def wrapper():
        """Wrapper function docstring"""
        print("Before call")
        func()
        print("After call")
    return wrapper

@decorator
def say_hello():
    """This function says hello"""
    print("Hello!")

print(say_hello.__name__)  
print(say_hello.__doc__)
```

### Fix: Using functools.wraps

To preserve the original function’s metadata, we use @functools.wraps(func) inside the decorator.

Using functools.wraps(), we keep the original function’s name and docstring, which is important for proper documentation and debugging.

```python
import functools

def decorator(func):
    @functools.wraps(func)
    def wrapper():
        """Wrapper function docstring"""
        print("Before call")
        func()
        print("After call")
    return wrapper

@decorator
def say_hello():
    """This function says hello"""
    print("Hello!")

print(say_hello.__name__) 
print(say_hello.__doc__)  
```

### Reusing Decorator in Python
A decorator in Python is like a regular Python function, so we can reuse it by recalling that specific decorator function. Here is an example of a decorator being used in different functions. 

Example:

```python
def show_steps(func):
    def wrapper(a, b):
        print(f"Calculating: {a} and {b}")
        result = func(a, b)
        print(f"Result is: {result}")
        return result
    return wrapper

# Use the decorator for multiplication
@show_steps
def multiply(a, b):
    return a * b

# Use the same decorator for squaring a number
@show_steps
def power(a, b):
    return a ** b

# Call the decorated functions
multiply(4, 5)
power(2, 3)

#In this example, we create a decorator show_steps to print inputs and results. We reuse it with both multiply and power functions to show calculation steps without repeating code
```

### Returning Values from Decorated Functions
We can control the return type from a decorated function. Here is an example showing what happens to the returned value from decorated functions. 

Example:

```python
# A simple decorator that modifies and returns the result
def double_result(func):
    def wrapper(a, b):
        result = func(a, b)
        return result * 2  # Modify the result before returning
    return wrapper

@double_result
def multiply(a, b):
    return a * b

# Call the decorated function
output = multiply(3, 4)
print("Final result:", output)
```

### Chaining Decorators in Python
Chaining decorators in Python means applying multiple decorators to a single function. The process involved placing several decorators one after the other, where the innermost decorator is used first. 

Example:

```python
import functools

# First decorator: doubles the return value
def double_return_value(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result * 2
    return wrapper

# Second decorator: squares the return value
def square_return_value(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result ** 2
    return wrapper

# Apply decorators (bottom one runs first)
@double_return_value      # runs second
@square_return_value      # runs first
def get_number():
    """Returns the number 5."""
    return 5

# Call the decorated function
result = get_number()
print(result)
```

### Python Decorators with Arguments
We can pass arguments to decorators by defining the decorator within another function that takes the arguments. After this, we use those arguments within the decorator. Also, we return it from the enclosing function. 

Example:

```python
import functools

# Outer function to accept arguments for the decorator
def repeat(n):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

# Apply the decorator with an argument
@repeat(3)
def greet(name):
    print(f"Hello, {name}!")

# Call the decorated function
greet("Alex")
```

### Fancy Decorators in Python
Now that you know how to use Python decorators on Functions, let’s see how to use decorators with classes. These decorators are called fancy decorators, and we can achieve them through the following two methods:

1. Decorating the methods of a class

2. Decorating a complete class

<b>1. Decorating the Methods of a Class</b>

There are various built-in decorators in Python that we can use with methods of a class:

- The first is @classmethod, which is used to create methods bound to the class and not the object. The class is passed as the first parameter to the class method used as a factory method to create specific instances of a class. 

- Next is @staticmethodStatic. It can't change the class or object state because there is no access to cls or self. It is only a part of a class namespace.

- The last one is @property, which we use to create getters and setters in Python for class attributes.

Here is an Example of three decorators in Python

1. @classmethod

```python
class Student:
    count = 0  # Class variable

    def __init__(self, name):
        self.name = name
        Student.count += 1

    @classmethod
    def total_students(cls):
        return cls.count

# Create students
s1 = Student("Amit")
s2 = Student("Sara")

# Call class method
print("Total students:", Student.total_students())
```

#### 2. @staticmethod

```python
class Calculator:
    
    @staticmethod
    def add(x, y):
        return x + y

# Call static method without creating object
print("Sum:", Calculator.add(10, 5))
```

#### 3. @property

```python
class Rectangle:
    def __init__(self, width, height):
        self._width = width
        self._height = height

    @property
    def area(self):
        return self._width * self._height

# Create rectangle
rect = Rectangle(4, 5)

# Access area like a variable
print("Area:", rect.area)
```

### 2. Decorating a Complete Class
In Python, we can use decorators not only for functions but also for entire classes. When we decorate a class, the decorator receives the class itself as an argument and can modify or enhance how that class behaves, especially during object creation.

<b>Important:</b> This does not change the behavior of class methods directly. It changes what happens when we create an object from the class.

Example:

```python
# Class decorator that logs when an object is created
def log_class(cls):
    class NewClass(cls):
        def __init__(self, *args, **kwargs):
            print(f"Creating object of class: {cls.__name__}")
            super().__init__(*args, **kwargs)
    return NewClass

# Apply the decorator to the class
@log_class
class Student:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(f"Student Name: {self.name}")

# Create objects of the decorated class
s1 = Student("Aman")
s1.show()
```

### Classes as Decorators in Python
In Python, we can also use classes as decorators, not just functions. This is useful when you want to store some state or run logic multiple times with more control.

To make a class work as a decorator:

It must have an `__init__`() method to take the function as an argument.

It must also have a `__call__`() method so it can be used like a function.

Use functools.update_wrapper() to keep the original function’s name and docstring.

<b>Example: Count How Many Times a Function is Called</b>

```python
import functools

# Create a class-based decorator
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.count = 0
        functools.update_wrapper(self, func)

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Call {self.count} of {self.func.__name__}")
        return self.func(*args, **kwargs)

# Apply the class as a decorator
@CountCalls
def greet(name):
    print(f"Hello, {name}!")

# Call the decorated function multiple times
greet("Alex")
greet("Sara")
greet("Charlie")
```

## Closure in Python

### Introduction

A Python closure is when we create a function inside another function and return it. The inner function remembers the values from the one, even after it finishes running.

Syntax:

```python
def outer_func():
    outer_var = "value"

    def inner_func():
        print(outer_var)

    return inner_func
```

### In this syntax, we:
- Create an outer function with a variable

- Write an inner function that uses the outer variable

- Return the inner function so it can be used later as a closure

## Nested Functions in Python

In Python, we can create a function inside another function, but unlike closures, it does not return the inner function.

Example:

```python
def greet():
    # Outer function
    def message():
        # Inner function
        print("Hello from the inner function!")
    message()  # Calling inner function

greet()
```

In this example, we create an inner function inside an outer one and call it to print a simple greeting message.

<b>Note:</b> The nested function runs only when the outer function runs and doesn’t remember values after it ends. A closure returns the inner function and keeps outer values saved for later use.

## What Is a Closure in Python?

In Python, a closure happens when a nested function remembers and uses variables from the outer function, even after the outer function has finished running. This allows the inner function to access those variables later.

A Python closure helps us keep certain data private or create function-based wrappers without using classes. It’s often used in decorators and callback functions.

<b>Key Points to Remember:</b>

- Inner function accesses outer variables.

- Outer function returns inner function.

- Data is remembered in memory.

- Common in decorators and callbacks.

Example:

```python
def greet(name):
    message = f"Hello, {name}!"

    def say():
        print(message)  # Inner function uses 'message' from outer scope

    return say  # Returning the inner function

# Creating a closure
greeting = greet("Sam")

# Calling the closure
greeting()
```

In this example, we create a function inside another and return the inner one. It uses the message variable from the outer function.

When we call greeting(), it prints the saved message: "Hello, Sam!".

### When and Why to Use Python Closures?

We use a closure in Python when we want an inner function to remember values from the outer function. It helps keep things clean, private, and easy to reuse later.

<b>Use Closures When:</b>
- We need to remember outer values

- We want to hide data

- We avoid using classes for small tasks

- We need custom behavior (like decorators)

Example:

```python
def multiplier(factor):
    # Outer function takes a number as 'factor'
    def multiply(number):
        # Inner function uses 'factor' from outer scope
        return number * factor
    return multiply  # Returning the inner function

# Creating a closure
double = multiplier(2)

# Calling the closure
result = double(5)
print("Result:", result)
```

## Property in Python: Method and Decorator

In Python, we often want to control how class attributes are accessed or modified. Instead of using direct access, we use special methods to manage the behavior.

The Python property() function helps us link these methods (getter, setter, deleter) to an attribute, making the code cleaner and more secure.

### What is property() in Python?

In Python, we use the Python property() function to manage how we access or modify class attributes. It helps us add logic using get, set, and del methods while working with object data.

With Python property(), we can control data like a method but access it like a variable.

Syntax:

property(fget=None, fset=None, fdel=None, doc=None)

<b>Parameters</b>

- fget() is an optional argument used to get the value of an attribute. Its default value is None.

- fset() is another optional argument used to set the value of an attribute. 

- fdel() is an optional argument used to delete an attribute’s value. 

- doc(), or docstring, contains the documentation for the attribute of the class.

### Reture

it returns the propert attribute from the getter, setter, and deleter.

<b>Notes</b>

- In case of no arguments, the property() method returns the base property attribute that doesn’t include any getter, setter, or deleter.

- In case there is no doc, the property() method takes the docstring of the getter.

## Creating properties in Python

There are two ways to create a property for a class in Python. They are as follows:

1. Using property() method

2. Using @property decorator

Let’s look at both with simple examples.

<b>Using property() Method</b>
 
We manually pass getter, setter, and deleter methods to the property() function.

Example:

```python
class Person:
    def __init__(self):
        self._age = 0

    def get_age(self):           # Getter method
        return self._age

    def set_age(self, value):    # Setter method
        if value >= 0:
            self._age = value
        else:
            print("Invalid age")

    def del_age(self):           # Deleter method
        del self._age

    age = property(get_age, set_age, del_age)  # Creating property

p = Person()
p.age = 30        # Calls set_age()
print(p.age)      # Calls get_age()
```

### Using @property Decorator

We use Python decorators to define a getter, setter, and deleter in a cleaner and more concise manner.

Example:

```python
class Person:
    def __init__(self):
        self._age = 0

    @property
    def age(self):         # Getter method
        return self._age

    @age.setter
    def age(self, value):  # Setter method
        if value >= 0:
            self._age = value
        else:
            print("Invalid age")

    @age.deleter
    def age(self):         # Deleter method
        del self._age

p = Person()
p.age = 25         # Calls setter
print(p.age)       # Calls getter
del p.age          # Calls deleter
```

### Python Property vs Attribute

In Python, we use attributes to store values directly. But when we need control, such as validation or calculated values, we use properties. Properties allow us to manage how attributes are accessed or modified.

<b>Key Differences</b>

| **Aspect**      | **Attribute**               | **Property**                                   |
| --------------- | --------------------------- | ---------------------------------------------- |
| **Access**      | Direct access to a variable | Accessed through methods (`getter` / `setter`) |
| **Validation**  | Not possible                | Easily done inside setter                      |
| **Syntax**      | Direct access (`self.x`)    | Uses `@property` decorator or `property()`     |
| **Behavior**    | Static (fixed value)        | Dynamic (can compute or validate on access)    |
| **Flexibility** | Less flexible               | More flexible and safer                        |

Example 1: Using a Regular Attribute

```python
# Regular attribute: No control or validation

class Car:
    def __init__(self):
        self.speed = 100  # Direct assignment

# Creating object
c1 = Car()
print("Car Speed:", c1.speed)  
```

```python
# Example 2: Using a Property with Getter and Setter

# Using property: Adds control with validation

class Vehicle:
    def __init__(self):
        self._speed = 0

    @property
    def speed(self):
        return self._speed

    @speed.setter
    def speed(self, value):
        if value < 0:
            raise ValueError("Speed cannot be negative")
        self._speed = value

# Creating object
c2 = Vehicle()
c2.speed = 80
print("Vehicle Speed:", c2.speed) 
```

## Property Setter in Python

The property setter in Python allows us to control how a value is assigned to an attribute. It adds validation or extra logic when setting a value, rather than just storing it directly.

We use the @property_name.setter decorator to define a setter method. This keeps our code clean while adding control to our class attributes.

Example:

```python
# Class with a property setter
class Product:
    def __init__(self):
        self._price = 0

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value

p = Product()
p.price = 150
print("Product Price:", p.price)
```

## Property Deleter in Python

Just like we use a setter to update a value, we can use a deleter to delete a property in Python. After deletion, we can’t access that property again using the same object.

To define a deleter, we use @property_name.deleter. Then we use the del keyword, and Python runs the delete logic we define.

Example:

```python
# Class with a property deleter
class Account:
    def __init__(self):
        self._balance = 1000

    @property
    def balance(self):
        return self._balance

    @balance.deleter
    def balance(self):
        print("Deleting balance...")
        del self._balance

a = Account()
print("Balance:", a.balance)  # Output: 1000
del a.balance                 # Output: Deleting balance...
```
