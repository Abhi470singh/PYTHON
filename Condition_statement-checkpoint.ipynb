## Conditional Statements
<b>🧠 Definition:</b>Conditionals in Python allow us to run code only when a specific condition is true. They help us make decisions and control how our program flows based on comparison, values or logical conditions.

- A conditional statement checks whether a condition (expression) is True or False.
- Based on the result, Python decides which block of code to execute.

##### Why do I need conditional statements in Python?
You use conditional statements when you want your program to choose between actions. For example, you can run one block if something is true and another if it’s false, just like real-life decisions.

```python
# for Exaple: Python program to check voting eligibility
age = 20
if age >= 18:
    print("You are eligible to vote.")
```

#### Python Conditional Operators for Decision Making
 We use different mathematical logic in Python to implement if-else statements and check conditions in various situations. Here are the operators we use with Python conditional statements:

-  Equals (==): a == b
- Not Equals (!=): a!=b
- Greater Than (>): a > b
- Less Than (<): a < b
- Greater Than or Equal to: a >= b
- Less Than or Equal to: a <= b

#### Python if Conditionnal Statement

We use the if statement in python to run a block of code only when a condition is true. it helps us controls the program flow based on conditions.

<b> Syntax:</b>

if condition:
    # code block

The flow of Python conditional statements programs is determined by the test expression. First, it checks the test expression. If it is true, the program moves to the if statement code block and executes it. If the test expression is False, the program jumps out of the if statement.

```python
age = 18
if age >= 18:
    print("You are eligible to vote.")
```

#### Python if-else Conditional Statement
The Python if statement executes a block of code only when the given condition is true; if it is false, it doesn’t execute anything. So, the if-else statement addresses this issue by merging the else statement, which is executed when the condition is false. 

So, if the condition is true, the program executes the if block of code; if the condition is false, it executes the if-else code block. 

<b> Syntax:</b>
if condition:
    # code block if condition is true
else:
    # code block if condition is false

<b> Flowchart of the if-else Statement</b>

![alt text](<flowchart-of-the-if-else-statement (2).webp>)

The program checks the test expression. If it is true, the statement in the if block will be executed based on indentation. If the test expression is false, the statement in the else block will be executed. 

```python
language = "Python"

if language == "Python":
    print("You are coding in Python")
else:
    print("You are using another language")
```

#### Python Nested if Conditional Statement
Nested if statements allow us to place one if statement inside another. This helps when we need to check multiple conditions in a structured way.

We use conditional statements in Python to control the flow of code based on more than one condition.

<b>Syntax:</b>
if condition1:
    if condition2:
        # code to run

###### Flowchart of the Nested If Statement
![alt text](flowchart-of-python-nested-if-statement.webp)

The given flowchart shows that the flow of a program is determined by the test expression. If the condition is true, it tests the nested test condition. If that condition is also true, it executes the nested if statements. 

However, if the condition is false, it executes the nested else statements. When the first condition is false, the program executes the else statement below it.

<b> Example:</b>

```python
language = "Python"
experience = 3

if language == "Python":
    print("You are using Python")
    if experience >= 2:
        print("You are an experienced Python developer")
else:
    print("Please choose Python to continue")
```

#### Python if-elif-else Conditional Statement
The if-elif-else statement helps us check multiple conditions one by one. Once a true condition is found, its block runs, and the rest are skipped.

We use these statements to handle different conditions in Python, making decision-making in programs flexible and clear.

<b> Syntax:</b>

if condition1:

        # code block

elif condition2:

        # code block

else:

        # code block

##### Flowchart of the elif Statement in Python
![alt text](<flowchart-of-elif-statement-in-python (1)-1.webp>)
As we can see, the program’s flow is based on multiple test expressions. If any of the given test expressions is true, the program executes the block of statements under that if or else statement. If none of the conditions are true, it executes the else block of the statement.

<b> Example:</b>

```python
score = 70

if score >= 90:
    print("Grade: A")
elif score >= 60:
    print("Grade: B")
else:
    print("Grade: C")
```

#### Python Ternary Expression
The ternary operator allows us to write a Python conditional statement in one line. It helps decide what value to return based on a condition, making the code shorter and cleaner.

Shorthand if in Python
We use this when we want to return a value if a condition is true and do nothing otherwise.

<b> Syntax:</b>
        if expression: statement

<b>Example:</b>

```python
score = 78
if score >= 33:print("Pass")
```

#### Shorthand if-else in Python
We use the if-else ternary expression to choose between two values depending on whether the condition is true or false.

<b>Syntax: </b>

value_if_true if condition else value_if_false

<b>Example:</b>

```python
marks = 75
result = "Pass" if marks >= 40 else "Fail"
print(result)
```

#### 1. Python Program to Find the Largest of Three Numbers

```python
# Program to find the largest of three numbers

# Taking input from the user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

# Compare the numbers
if (num1 >= num2) and (num1 >= num3):
    largest = num1
elif (num2 >= num1) and (num2 >= num3):
    largest = num2
else:
    largest = num3

# Display the result
print("The largest number is:", largest)
```

##### Using the max() Function (Shortcut)

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

print("The largest number is:", max(num1, num2, num3))
```

#### 2. Python Program to Check Leap Year

```python
# Program to check if a year is a leap year

year = int(input("Enter a year: "))

# Leap year conditions:
# 1. Divisible by 4
# 2. If divisible by 100, it must also be divisible by 400

if (year % 400 == 0) or ((year % 4 == 0) and (year % 100 != 0)):
    print(year, "is a Leap Year")
else:
    print(year, "is NOT a Leap Year")
```

🧠 Explanation:

- A year is a leap year if:

-   It is divisible by 4, but not by 100, OR

-   It is divisible by 400.

#### Python Program to Find HCF or GCD
✅ Python Program to Find HCF (or GCD)

HCF = Highest Common Factor
GCD = Greatest Common Divisor
(Both mean the same thing.)

🧮 Program 1: Using a Loop (Without Library)

✅ Program 2: Using math.gcd() Function (Easiest Way)

```python
# Program to find HCF (or GCD) of two numbers

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Find the smaller number
if num1 > num2:
    smaller = num2
else:
    smaller = num1

for i in range(1, smaller + 1):
    if (num1 % i == 0) and (num2 % i == 0):
        hcf = i

print("The HCF (GCD) of", num1, "and", num2, "is:", hcf)
```

```python
import math

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("The HCF (GCD) of", num1, "and", num2, "is:", math.gcd(num1, num2))
```

#### Python Program to Check Armstrong Number

✅ Program: Check Armstrong Number

✅ Program 2 (Short version using list comprehension):

```python
# Program to check if a number is an Armstrong number

num = int(input("Enter a number: "))

# Find the number of digits
order = len(str(num))

# Initialize sum
sum = 0

# Copy the original number
temp = num

while temp > 0:
    digit = temp % 10            # Get last digit
    sum += digit ** order        # Raise to the power of number of digits
    temp //= 10                  # Remove last digit

# Compare
if num == sum:
    print(num, "is an Armstrong number")
else:
    print(num, "is NOT an Armstrong number")
```

🧠 Explanation:

An Armstrong number (or narcissistic number) is a number that is equal to the sum of its own digits raised to the power of the number of digits.

| Number | Calculation              | Result      |
| ------ | ------------------------ | ----------- |
| 153    | 1³ + 5³ + 3³ = 153       | ✅ Armstrong |
| 370    | 3³ + 7³ + 0³ = 370       | ✅ Armstrong |
| 9474   | 9⁴ + 4⁴ + 7⁴ + 4⁴ = 9474 | ✅ Armstrong |

```python
num = int(input("Enter a number: "))
order = len(str(num))
if num == sum(int(digit) ** order for digit in str(num)):
    print(num, "is an Armstrong number")
else:
    print(num, "is NOT an Armstrong number")
```

```python
# Program to calculate Compound Interest

# Input from the user
principal = float(input("Enter the principal amount: "))
rate = float(input("Enter the rate of interest: "))
time = float(input("Enter the time (in years): "))

# Calculate compound interest
# Formula: CI = P * (1 + R/100) ** T - P
compound_interest = principal * ((1 + rate / 100) ** time) - principal

# Display result
print("Compound Interest = {:.2f}".format(compound_interest))
```
