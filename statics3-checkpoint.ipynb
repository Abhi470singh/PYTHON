### <span style="color:lime">1️⃣ Estimation (in Inferential Statistics)</span>

- Estimation is a statistical method used to predict or infer unknown population parameters (like mean, proportion, variance) using sample data.

👉 Since studying the entire population is often impossible, we take a sample and use it to estimate population values.

#### <span style="color:limeGreen">Types of Estimation</sapn>
##### <span style="color:PaleGreen">1️⃣ Point Estimation</sapn>

- A single numerical value is used to estimate a population parameter.

<b>Formulas:-</b>

- Population Mean estimate
$\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i$
	​
- Population Proportion estimate
$\hat{p} = \frac{x}{n}$

<b>Examples:</b>

- Sample mean (𝑥ˉ) → estimates population mean (μ)
- Sample proportion (𝑝^) → estimates population proportion (p)

<b>Example:</b>
- If the average score of a sample of students is 72, then 72 is the point estimate of the population mean.

<b>📌 Limitation:</b>
- Does not show uncertainty or accuracy of the estimate.

```python
import numpy as np

sales = [1200, 1500, 1300, 1400, 1600]

point_estimate = np.mean(sales)
print("Point Estimate (Mean):", point_estimate)
```

##### <span style="color:lightGreen">2️⃣ Interval Estimation (Confidence Interval)</span>
- An interval estimate gives a range of values where the true population parameter is likely to lie.

<b>Formula :-</b>

📌 Confidence Interval for Mean (σ known)

$$\bar{x} \pm Z_{\alpha/2}\,\frac{\sigma}{\sqrt{n}}$$

🔍 Where:

- 𝑥
ˉ = sample mean


- Z
α/2
= Z-value (e.g., 1.96 for 95% confidence)
- σ = population standard deviation
- n = sample size

##### Point vs Interval Estimation (Quick Recap)
| Feature     | Point Estimation | Interval Estimation |
| ----------- | ---------------- | ------------------- |
| Value       | Single number    | Range               |
| Uncertainty | ❌ No             | ✅ Yes               |
| Reliability | Low              | High                |
| Example     | Mean = 1400      | Mean ∈ (1300, 1500) |

```python
from scipy import stats

data = [1200, 1500, 1300, 1400, 1600]
mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

confidence = 0.95
z = stats.norm.ppf((1 + confidence) / 2)

margin_error = z * (std / np.sqrt(n))
ci = (mean - margin_error, mean + margin_error)

print("95% Confidence Interval:", ci)
```

### <span style="color:DarkGreen">1️⃣ Z-test</span>

Used when:

- Sample size n≥30
- Population standard deviation (σ) is known

Test statistic :

$$Z = \frac{\bar{x} - \mu}{\sigma/\sqrt{n}}$$

<b>🔹 Problem Statement</b>

- A company claims that the average weight of its product is 50 units.

- Population mean  μ=50

- Population standard deviation  σ=10

- Sample data: [52,55,48,51,54,53,49,50,56,52]

- Significance level  : α=0.05

👉 Question: Is the sample mean significantly different from 50?

```python
from scipy.stats import norm

# Sample data
sample = np.array([52, 55, 48, 51, 54, 53, 49, 50, 56, 52])

# Given values
population_mean = 50
population_std = 10
alpha = 0.05

# Calculations
sample_mean = np.mean(sample)
n = len(sample)

z_value = (sample_mean - population_mean) / (population_std / np.sqrt(n))

# Critical value (two-tailed)
z_critical = norm.ppf(1 - alpha/2)

# p-value
p_value = 2 * (1 - norm.cdf(abs(z_value)))

print("Sample Mean:", sample_mean)
print("Z value:", z_value)
print("Critical Z value:", z_critical)
print("P-value:", p_value)
```

```python
import numpy as np
import math
import matplotlib.pyplot as plt

# Sample data
sample = np.array([52, 55, 48, 51, 54, 53, 49, 50, 56, 52])

# Given values
mu = 50
sigma = 10
alpha = 0.05

# Calculations
x_bar = np.mean(sample)
n = len(sample)

z = (x_bar - mu) / (sigma / math.sqrt(n))

# Critical Z value (two-tailed, alpha = 0.05)
z_critical = 1.96

print("Sample Mean:", x_bar)
print("Z value:", z)
print("Critical Z:", z_critical)
```

```python
x = np.linspace(-4, 4, 400)
pdf = (1 / math.sqrt(2 * math.pi)) * np.exp(-0.5 * x**2)

plt.figure()
plt.plot(x, pdf)
plt.axvline(z, linestyle='--', label='Calculated Z')
plt.axvline(z_critical, linestyle=':', label='Critical Z')
plt.axvline(-z_critical, linestyle=':')
plt.legend()
plt.xlabel("Z value")
plt.ylabel("Probability Density")
plt.title("Z-Test (Without scipy)")
plt.show()
```

##### <span style="color:YellowGreen">2️⃣ t-test</span>
<b>📌 Used when:</b>

- Sample size is small (n<30)
- Population standard deviation is unknown
- Data is approximately normally distributed

<b>📌 Types of t-test</b>

<sapn style="color:OliveDrab">1️⃣ One-sample t-test</sapn>

- Compares sample mean with a known population mean
$$t = \frac{\bar{x} - \mu}{s/\sqrt{n}}$$

Where:

- 𝑥
ˉ = sample mean
- μ = population mean
- s = sample standard deviation
- n = sample size

<span style="color:Olive">2️⃣ Independent (Two-sample) t-test</span>

- Compares means of two independent groups

<b>Test statistic (equal variance):</b>
$$t = \frac{\bar{x}_1 - \bar{x}_2}{s_p\sqrt{\frac{1}{n_1}+\frac{1}{n_2}}}$$

<b>Pooled variance:</b>

$$s_p^2 = \frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}$$

<span style="color:DarkOliveGreen">3️⃣ Paired t-test</span>

- Compares means of related/paired observations

<b>Test statistic:</b>
$$t = \frac{\bar{d}}{s_d/\sqrt{n}}$$


```python
from scipy import stats

# Sample data
sample = np.array([52, 55, 48, 51, 54, 53, 49, 50, 56, 52])

# Hypothesized mean
mu = 50

# One-sample t-test
t_stat, p_value = stats.ttest_1samp(sample, mu)

# Degrees of freedom
df = len(sample) - 1

# Critical t-value (two-tailed, alpha=0.05)
alpha = 0.05
t_critical = stats.t.ppf(1 - alpha/2, df)

print("Sample mean:", np.mean(sample))
print("t-statistic:", t_stat)
print("p-value:", p_value)
print("Critical t-value:", t_critical)
```

```python
# t-distribution plot
x = np.linspace(-4, 4, 400)
y = stats.t.pdf(x, df)

plt.figure()
plt.plot(x, y, label='t-distribution (df=9)')
plt.axvline(t_stat, linestyle='--', color='red', label='Calculated t')
plt.axvline(t_critical, linestyle=':', color='green', label='Critical t')
plt.axvline(-t_critical, linestyle=':', color='green')
plt.xlabel("t value")
plt.ylabel("Probability Density")
plt.title("One-Sample T-Test")
plt.legend()
plt.show()
```

### 3️⃣ Chi-square (χ²) Test
<b>📌 Used for:</b>

- Categorical data

Testing:

- Independence (association between two categorical variables)
- Goodness of fit (whether observed frequencies match expected frequencies)

<b>📌 Test Statistic</b>

<span style="color:MediumAquamarine">1️⃣ Chi-square formula</span>
$$\chi^2 = \sum \frac{(O_i - E_i)^2}{E_i}$$

Where:

- 𝑂𝑖= Observed frequency

- 𝐸𝑖= Expected frequency

```python
import pandas as pd
import scipy.stats as stats
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data: Survey of 100 people, checking relationship between Gender and Preference
data = {'Gender': ['Male','Male','Male','Female','Female','Female','Male','Female','Male','Female'],
        'Preference': ['A','B','A','A','B','B','A','B','B','A']}

df = pd.DataFrame(data)

# Create a contingency table
contingency_table = pd.crosstab(df['Gender'], df['Preference'])
print("Contingency Table:\n", contingency_table)

# Chi-square test
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)
print(f"\nChi2 Statistic = {chi2}")
print(f"P-value = {p}")
print(f"Degrees of Freedom = {dof}")
print("Expected Frequencies:\n", expected)

# Decision
alpha = 0.05
if p < alpha:
    print("Reject H0: There is an association between Gender and Preference")
else:
    print("Fail to Reject H0: No association between Gender and Preference")

# Visualization
sns.heatmap(contingency_table, annot=True, cmap='YlGnBu', fmt='d')
plt.title('Contingency Table Heatmap')
plt.show()
```

### <span style="color:DarkSeaGreen">5️⃣ ANOVA (Analysis of Variance)</span>
<b>📌 Question:</b>

- Do three different teaching methods have different mean scores?

<b>📌 Hypotheses:</b>

- <b>Null hypothesis (H₀):</b>

$H_0$: \mu_1 = \mu_2 = \mu_3$
(All means are equal)

- <b>Alternative hypothesis (H₁):</b>

$H_1$: \text{At least one mean is different}$

```python

# Sample data: Scores of students with 3 different teaching methods
data = {
    'Method_A': [85, 90, 78, 92, 88],
    'Method_B': [80, 82, 79, 85, 87],
    'Method_C': [88, 90, 92, 91, 89]
}

# Convert to DataFrame
df = pd.DataFrame(data)

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(df['Method_A'], df['Method_B'], df['Method_C'])

print(f"F-statistic = {f_stat}")
print(f"P-value = {p_value}")

# Decision at alpha = 0.05
alpha = 0.05
if p_value < alpha:
    print("Reject H0: At least one teaching method has a different mean")
else:
    print("Fail to Reject H0: No significant difference between teaching methods")

# Visualization
sns.boxplot(data=df)
plt.title('Comparison of Scores Across Teaching Methods')
plt.ylabel('Scores')
plt.show()
```
