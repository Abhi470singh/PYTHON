# <span style="color:yellow">Iterators in Python</span>

Iterators in Python allow us to access items from collections like lists or strings, one by one. We use iter() to create the iterator and next() to fetch each item.

Example:

```python
fruits = ["apple", "mango", "cherry"]
it = iter(fruits)

print(next(it))  # apple
print(next(it))  # mango
print(next(it))  # cherry
```

## <span style="color:green">What is an Iterators in Python?</spam>

<b>Iteration</b> refers to the process of taking each item from a collection or sequence, one after another. Whenever you use a loop (either explicitly or implicitly) to go through all the elements in a data structure, it is considered iteration.

In Python, the most common way of performing iteration is using loops like <b>for and while</b> , which allow you to access each element of an iterable, such as lists, tuples, or Strings.

<b>Types of Iterables:</b>
1. <b>Lists </b>– An ordered collection of items.
2. <b>Tuples</b> – Similar to lists but immutable.
3. <b>Dictionaries</b> – Collection of key-value pairs.
4. <b>Strings</b> – A sequence of characters.
5. <b>Sets</b> – An unordered collection of unique items.

Example:



```python
# Example of iteration over a list
my_list = [1, 2, 3, 4]
for item in my_list:
 print(item)
```

### Two Primary Methods of Python Iterators:
-  `__iter__`(): Initializes and returns the iterator object.

- `__next__`(): Returns the next item and raises StopIteration when no items are left.

Here is a Python iterator with proper syntax and an example:

<span style="color:yellow;">`__iter__`()</span>

This method initializes an iterator and returns the iterator object itself. It allows an object to be used in a for loop or with the next() function.

Syntax:

myiterator = iter(iterable)

```python
my_list = [10, 20, 30]
myiterator = iter(my_list)
print(type(myiterator))   # Checking the type of the iterator
```

`__next__`()

This method returns the next element in the collection. If no more elements are left, it raises a StopIteration error. The for loop in Python automatically calls the `__next__`() method to get each item during iteration.

Syntax:

item = next(myiterator)

```python
my_list = [10, 20, 30]
myiterator = iter(my_list)

print(next(myiterator))   # 10
print(next(myiterator))   # 20
print(next(myiterator))   # 30
# print(next(myiterator))   # Uncommenting this line raises StopIteration
```

## <span style="color:lime">Iterator vs Iterable in Python</span>

In Python, iterables are objects like lists, tuples, or strings that can be looped over. These objects have the `__iter__`() method, which returns an iterator.

Iterators in Python are objects that keep track of the current position while looping. They use the `__next__`() method to return items one at a time and raise StopIteration when done.

## <span style="color:Orange;">Iterating Through an Iterator in Python </span>

An iterator in Python allows us to loop through a sequence of values using the next() function. Iterators are created from iterables like lists, tuples, or custom classes.

<b>Using next() Manually</b>

We can use the next() function to access items one by one until the sequence ends.

Example:

```python
# Define an iterable (a list)
numbers = [1, 2, 3]

# Create an iterator using iter()
iterator = iter(numbers)

# Access items using next()
print(next(iterator))  
print(next(iterator)) 
print(next(iterator))  
```

#### <span style="color:navy">Using a for Loop</span>

We often use a for loop to automatically iterate through all elements in an iterator.

Example:

```python
# Define a list
numbers = [1, 2, 3]

# Use a for loop to iterate
for num in numbers:
    print(num)
```

```python
#example:
class CountUpto:
    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):
        return self  # The class itself is the iterator

    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        else:
            raise StopIteration

# Create an iterator object
counter = CountUpto(3)

# Iterate using next()
print(next(counter))  # Output: 1
print(next(counter))  # Output: 2
print(next(counter))  # Output: 3
# print(next(counter))  # Raises StopIteration
```

### <span style="color:indigo">StopIteration in Python</span>

In iterators in Python, StopIteration is an error that tells the loop or program there are no more items left to fetch from the iterator.

Syntax:

raise StopIteration

```python
class Counter:
    def __init__(self, max):
        self.num = 1
        self.max = max

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= self.max:
            val = self.num
            self.num += 1
            return val
        else:
            raise StopIteration

# Create object
counter = Counter(3)

for num in counter:
    print(num)
```

### <span style="color:teal">Python Infinite Iterators</span>

Python infinite iterators generate endless values without stopping. We use them when we want continuous data flow, often with tools like itertools.count() or custom loops.

Example:

```python
import itertools

# Create an infinite iterator starting from 1
counter = itertools.count(1)

for num in counter:
    print(num)
    if num == 5:
        break
```
