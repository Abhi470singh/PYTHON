## Inheritance in Python
- Python inheritance allows us to create a new class that uses the features of an existing class. it helps us reuse code and build relationships between classes.

<b>Syntax:</b>

<span style="color:red;">class</span> <span style="color:blue;">Parent:</span>

        # parent class code

<span style="color:red;">class</span> <span style="color:black;">Child</span> <span style="color:blue;">(Parent)</span>

Example:

```python
class Person:
    def show_name(self):
        print("Name: Alex")

class Student(Person):
    def show_course(self):
        print("Course: Python Programming")

# Creating object of Student class
s = Student()
s.show_name()    # Inherited from Person
s.show_course()  # Defined in Student
```

### <span style="color:#FFD700;">Single Inheritance</span> in Python

Single inheritance in Python means one class inherits from one parent class. We use it when we want a child class to reuse or extend the functionality of a single base class.


Example:

```python
# Example of Single Inheritance in Python
class Person:
    def display_name(self):
        print("Name: Sara")

class Student(Person):
    def display_course(self):
        print("Course: Python Programming")

# Using the child class
s = Student()
s.display_name()   # Inherited from Person
s.display_course() # Defined in Student
```

```python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
```

### <span style="color:blue;">Multiple Inheritance</span> in Python

In multiple inheritance, a single subclass inherits from multiple superclasses. So, the child class can access attributes and methods from two or more parent classes. 

However, if the parent classes have methods with the same name, the child class uses the method from the first parent in the order they are listed.


Example:

```python
# Parent class 1
class Teacher:
    def teach(self):
        print("Teaches Math and Science")

# Parent class 2
class Coach:
    def train(self):
        print("Trains in Football and Athletics")

# Child class inherits from both Teacher and Coach
class Student(Teacher, Coach):
    def activities(self):
        print("Student is active in both academics and sports")
        self.teach()  # Calling method from Teacher
        self.train()  # Calling method from Coach

# Create object of Student
s = Student()
s.activities()
```

```python
# Multiple
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def review(self):
        print ("Customer review")

class SmartPhone(Phone, Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

### <span style="color:green;">Multilevel Inheritance</span> in Python

Python multilevel inheritance means a class inherits from a child class, which itself inherits from another parent class. This forms a chain of inheritance, passing features from one level to the next.

With multilevel inheritance in Python, we can build step-by-step class hierarchies, allowing each level to extend or reuse the previous one.


Example:

```python
# Base class
class Grandparent:
    def show_grandparent(self):
        print("Grandparent: Shares wisdom")

# Intermediate class inheriting from Grandparent
class Parent(Grandparent):
    def show_parent(self):
        print("Parent: Provides guidance")

# Child class inheriting from Parent
class Child(Parent):
    def show_child(self):
        print("Child: Learns and grows")

# Create object of Child class
c = Child()
c.show_grandparent()  # Inherited from Grandparent
c.show_parent()       # Inherited from Parent
c.show_child()        # Defined in Child
```

```python
# multilevel
class Product:
    def review(self):
        print ("Product customer review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
s.review()
```

### <span style="color:#D81B60;">Hierarchical Inheritance</span> in Python

Hierarchical inheritance is the opposite of multiple inheritance. Therefore, more than one child class can be derived from a single-parent class.


Example:

```python
# Parent class
class Animal:
    def sound(self):
        print("Animals make different sounds")

# Child class 1
class Dog(Animal):
    def bark(self):
        print("Dog: Barks")

# Child class 2
class Cat(Animal):
    def meow(self):
        print("Cat: Meows")

# Creating objects of both child classes
d = Dog()
d.sound()   # Inherited from Animal
d.bark()    # Defined in Dog

c = Cat()
c.sound()   # Inherited from Animal
c.meow()    # Defined in Cat
```

```python
# Hierarchical
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    pass

class FeaturePhone(Phone):
    pass

SmartPhone(1000,"Apple","13px").buy()
FeaturePhone(10,"Lava","1px").buy()
```

### <span style="color:black;">Hybrid Inheritance</span> in Python

Hybrid inheritance combines two or more types of inheritance. Hence, we can see multiple relationships between parent and child classes across different levels. We can say that hybrid inheritance in Python is a mixture of more than one inheritance style, like single, multiple, or multilevel.


Note:  In hybrid inheritance, we use Python’s MRO (Method Resolution Order) to decide which parent’s method runs first without confusion.

Example:

```python
# Base class
class Vehicle:
    def show_type(self):
        print("Vehicle: Used for transportation")

# First level child (single inheritance)
class Car(Vehicle):
    def car_features(self):
        print("Car: Has 4 wheels and AC")

# Another base class for multiple inheritance
class Electric:
    def battery_info(self):
        print("Electric: Runs on battery")

# Hybrid class (inherits from Car and Electric)
class ElectricCar(Car, Electric):
    def features(self):
        print("ElectricCar: Eco-friendly and silent")

# Create object of ElectricCar
ecar = ElectricCar()
ecar.show_type()       # Inherited from Vehicle
ecar.car_features()    # Inherited from Car
ecar.battery_info()    # Inherited from Electric
ecar.features()        # Defined in ElectricCar
```

When we use the len() function, it works with different data types like strings, lists, tuples, sets, and dictionaries. However, it gives results based on the kind of data we are working with.

Polymorphism with Class Methods allows us to use the same method name across different classes, each with its own behavior. This keeps our code clean and adaptable.

```python
class Circle:
    def draw(self):
        print("Drawing a circle")

class Square:
    def draw(self):
        print("Drawing a square")

class Triangle:
    def draw(self):
        print("Drawing a triangle")

# List of shape objects
shapes = [Circle(), Square(), Triangle()]

# Call draw method on each shape
for shape in shapes:
    shape.draw()
```

#### Python Polymorphism With Inheritance
In Python, just like other languages, we use inheritance to allow child classes to inherit methods and attributes from a parent class. We can also change or customize these methods in the child class; this is called method overriding.

With Python polymorphism, we can access these overridden methods using the same method name, and each class responds based on its own version.

Example:

```python
# Parent Class
class Device:
    def __str__(self):
        return "Generic Device"

    def feature(self):
        return "Basic features"

# Child Class 1
class Laptop(Device):
    def feature(self):
        return "Laptop with keyboard and screen"

# Child Class 2
class Smartphone(Device):
    def specs(self):
        return "Touchscreen and camera"

# Main Program
a = Laptop()
b = Smartphone()

print(b)                  # Output: Generic Device
print(b.feature())        # Output: Basic features
print(a.feature())        # Output: Laptop with keyboard and screen
print(b.specs())          # Output: Touchscreen and camera
```

In this example, we create a parent class Device and two child classes. The Laptop class overrides the feature() method, while Smartphone inherits it and adds its own specs() method.

We use the same method name feature() in different classes, showing how Python polymorphism works with inheritance and method overriding.

### 1. Operator Overloading in Python
Operator Overloading in Python means we can define how built-in operators like +, -, *, and others behave for user-defined objects (like classes).

Instead of using these operators only with basic data types, we can make them work with our own classes by writing special methods such as __add__() for +, __sub__() for -, and so on.

Example:

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return Number(self.value - other.value)

    def __str__(self):
        return str(self.value)

n1 = Number(15)
n2 = Number(5)
print(n1 - n2) 
```

### How to Overload the Operators in Python ?
We overload the operators in Python by defining special methods inside a class. These methods start and end with double underscores, like __add__, __sub__, __mul__, etc.

- Create a class with necessary attributes.

- Add special methods like __add__, __sub__, __eq__, etc.

- Write logic to return a result or new object.

Example:

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

# Creating objects
p1 = Point(3, 4)
p2 = Point(1, 2)

# Using overloaded +
result = p1 + p2
print(result)  # Output: (4, 6)
```

### Overloading the Binary + Operator in Python

In Python, we use the __add__() method to Overload the binary + operator. it lets us define how objects of custom class should be added together.

Example:

```python
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    # Overloading the binary + operator
    def __add__(self, other):
        real_part = self.real + other.real
        imag_part = self.imag + other.imag
        return ComplexNumber(real_part, imag_part)

    def __str__(self):
        return f"{self.real} + {self.imag}i"

# Creating two complex number objects
c1 = ComplexNumber(3, 2)
c2 = ComplexNumber(1, 7)

# Adding two complex numbers using the overloaded + operator
c3 = c1 + c2

print(c3)  
```

### How Does Python Operator Overloading Work?
When we change the behavior of any existing operator using the process of operator overloading in Python, it redefines the special function automatically invoked when we use that operator with the objects. 

As we use operator overloading with a user-defined object, Python calls the relevant magic method to perform the operation.

<b>Example 1: Multiplying an Object with a Scalar</b>

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __rmul__(self, scalar):
        return self.__mul__(scalar)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v = Vector(2, 3)
v2 = v * 4    # Calls __mul__
v3 = 3 * v    # Calls __rmul__

print(v2)     
print(v3)   
```

<b>Example 2: String Representation</b>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"{self.title} by {self.author}"

    def __repr__(self):
        return f"Book('{self.title}', '{self.author}')"

book = Book("Learning Python", "Mark Lutz")

print(str(book))   
print(repr(book))
```

### Overloading Assignment Operators in Python
In Python, we can’t directly overload assignment operators like =, but we can overload compound assignment operators such as +=, -=, or *= using special methods like __iadd__(), __isub__(), etc., inside a class.

<b>Example: Overloading += Operator

```python
class Counter:
    def __init__(self, value):
        self.value = value

    def __iadd__(self, other):
        self.value += other
        return self

    def __str__(self):
        return str(self.value)

c = Counter(5)
c += 3
print(c)
```

### Overloading Comparison Operators in Python
In Python, we can overload comparison operators using special methods to compare custom objects. These include:

- == → __eq__()

- != → __ne__()

- > → __gt__()

- < → __lt__()

- >= → __ge__()

- <= → __le__()

Python Operator Overloading allows us to define how these symbols behave when used with user-defined classes.

Example:

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks

    def __ne__(self, other):
        return self.marks != other.marks

    def __gt__(self, other):
        return self.marks > other.marks

    def __lt__(self, other):
        return self.marks < other.marks

    def __ge__(self, other):
        return self.marks >= other.marks

    def __le__(self, other):
        return self.marks <= other.marks

s1 = Student(85)
s2 = Student(90)

print(s1 == s2)  # False
print(s1 != s2)  # True
print(s1 > s2)   # False
print(s1 < s2)   # True
print(s1 >= s2)  # False
print(s1 <= s2)  # True
```

### Overloading Unary Operators in Python
We can overload unary operators like -, +, and ~ in Python using special methods such as __neg__(), __pos__(), and __invert__() inside a class.

Example: Overloading ~ Operator

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __neg__(self):
        return Number(-self.value)

    def __str__(self):
        return str(self.value)

n = Number(10)
print(-n)  # Output: -10
```

### Method Overloading in Python
Method Overloading in Python means using one method name that performs different tasks depending on the number or type of arguments passed to it.

- Python Method Overloading allows a method to behave differently with different inputs.

- Python handles this using techniques like default arguments, *args, or **kwargs.

- It makes code more flexible, readable, and reduces repetition.

#### Python Method Overloading Example
Python Method Overloading allows us to define a single method that behaves differently based on the arguments we pass. While Python doesn't support it directly, we can still achieve similar behavior using smart techniques.

<b>We can achieve it by using:</b>
- Default Arguments: Set default values to handle missing arguments.

- *args: Accept multiple positional arguments.

- **kwargs: Accept multiple keyword arguments.

- Conditional Logic: Use statements to change method behavior based on arguments.

- Multiple Dispatch: Use decorators to overload methods by type.

Now, understand Python Method Overloading by some useful examples:

#### 1. Using Default Arguments 
Using default arguments means we assign default values to parameters, so the method can be called with or without passing all the arguments.

Example:

```python
# Example of method overloading using default arguments
class Greet:
    def hello(self, name="Guest"):
        print(f"Hello, {name}!")

# Creating object
g = Greet()
g.hello()          # Output: Hello, Guest!
g.hello("Amit")    # Output: Hello, Amit!
```

#### 2. Using Variable Arguments (*args)
We use *args to pass multiple values to a method without fixing the number of arguments. This approach helps us achieve Python method overloading by handling different numbers of inputs.

Example:

```python
# Example of method overloading using *args
class Calculator:
    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        print("Product is:", result)

# Creating object
c = Calculator()
c.multiply(2, 3)            # Output: Product is: 6
c.multiply(2, 3, 4)         # Output: Product is: 24
```

#### 3. Using Keyword Argument (**kwargs)
We use **kwargs to accept any number of keyword arguments in a method. It helps us create flexible behavior similar to method overloading in Python.

Example:

```python
# Example of method overloading using **kwargs
class Student:
    def details(self, **kwargs):
        for key, value in kwargs.items():
            print(f"{key}: {value}")

# Creating object
s = Student()
s.details(name="Sara", grade="A")
```

#### Overloading Using Method Dispatching
Python doesn’t support method overloading directly, but we can use a module called multipledispatch to make it work. It allows us to create methods with the same name but different arguments.

```python
# Example of method overloading using multipledispatch
from multipledispatch import dispatch

class Calculator:
    
    @dispatch(int, int)
    def add(self, a, b):
        print("Sum of integers:", a + b)
    
    @dispatch(float, float)
    def add(self, a, b):
        print("Sum of floats:", a + b)

# Creating object
c = Calculator()
c.add(2, 3)         # Output: Sum of integers: 5
c.add(2.5, 3.5)     # Output: Sum of floats: 6.0
```

#### Using a Single Method with Conditional Logic
We can write one method and use if-else conditions inside it to change its behavior based on arguments. This is a simple way to simulate overloading.

This approach helps us achieve Python method overloading without using extra decorators or libraries.

Example:

```python
# Example of method overloading using conditional logic
class Area:
    def calculate(self, shape, a, b=0):
        if shape == "square":
            print("Area of square:", a * a)
        elif shape == "rectangle":
            print("Area of rectangle:", a * b)
        else:
            print("Unknown shape")

# Creating object
obj = Area()
obj.calculate("square", 4)         
obj.calculate("rectangle", 4, 5)
```

### 3. Method Overriding In Python
Method Overriding in Python means writing a method in the child class with the same name and parameters as a method in the parent class to change its behavior.

- It works only with inheritance.

- The child class replaces the parent class method.

- It helps in writing custom logic for specific child classes.

In this example,

- data1 and data2 are different properties of the Shape and Square classes.

- The method no_of_sides() is overridden in the Square class, meaning it replaces the version in Shape.

- color() is a new method added only in the Square class.

### Method Overriding in Python: Example


```python
class Shape:
    def no_of_sides(self):
        print("This shape has many sides")

class Square(Shape):
    def no_of_sides(self):
        print("A square has 4 sides")

# Creating objects
s1 = Shape()
s2 = Square()

# Calling the method
s1.no_of_sides()   # Parent class method
s2.no_of_sides()   # Child class overrides the method 
```

#### Features of Method Overriding in Python
Let’s take a look at the key features of method overriding in Python to help you understand the concept better:

- It works only with inheritance, where a child class inherits from a parent class.

- The child class overrides a method defined in the parent class with the same name and parameters.

- It allows us to change the behavior of a method without modifying the parent class.

- - It helps to avoid code duplication and makes the code cleaner.

The method in both parent and child classes must have the same signature (name and arguments).

#### Prerequisites for Python Method Overriding
Python's Method overriding has a few prerequisites that we have listed in this section:

- The method must share the same name as the parent class.

- It can’t occur in the class and requires a child class derived from the parent class. Therefore, it is considered a part of the concept of inheritance.

- The method must have the same number of parameters as the superclass.

#### Method Resolution Order in Python
In method resolution order, a Python program searches for a class in multiple inheritance. Here are a few points to consider:

- Multi-inheritance refers to a class derived from more than one parent class.

- It follows the depth-first search approach.

- It is crucial in programs using multi-inheritance.

- First, the attributes are searched for in the base class. If not found, the search moves to the immediate superclass, and the search continues following the same approach. 

<b>Now let us understand this through an example: Diamond Inheritance</b>

In diamond inheritance, class D inherits from two classes B and C, which both inherit from the same class A.

This can create confusion about which method to use if both B and C override a method from A. Python handles this using Method Resolution Order (MRO).

```python
# Parent class A
class A:
    def show(self):
        print("From class A")

# Class B inherits from A
class B(A):
    def show(self):
        print("From class B")

# Class C also inherits from A
class C(A):
    def show(self):
        print("From class C")

# Class D inherits from both B and C
class D(B, C):
    pass  

# Creating an object of class D
d = D()

# Calling the show() method
d.show()
```

#### Python Method Overriding with Multiple and Multilevel Inheritance
We use method overriding in Python when a child class wants to change the behavior of a method inherited from its parent class. This becomes more interesting in multiple or multilevel inheritance, where method resolution order matters.

#### Python Method Overriding in Multiple Inheritance
In multiple inheritance, a class inherits from more than one parent class. In the image, class D inherits from both B and C, so it can use features from both. Overriding happens when class D provides its own version of a method already present in B or C.

Example:

```python
class A:
    def show(self):
        return "Method from class A"

class B(A):
    def show(self):
        return "Method from class B"

class C(A):
    def show(self):
        return "Method from class C"

class D(B, C):
    def show(self):
        return "Method from class D"

# Creating object of class D
d = D()
print(d.show())

# Check method resolution order
print(D.mro())
```

#### Python Method Overriding in Multilevel Inheritance
In multilevel inheritance, classes are connected in a chain: one inherits from another. In the image, class 3 inherits from class 2, which inherits from class 1. A child or grandchild class can override methods from any parent class.

Example:

```python
class One:
    def show(self):
        return "Method from class One"

class Two(One):
    def show(self):
        return "Method from class Two"

class Three(Two):
    def show(self):
        # Using super() to call the parent method
        parent_msg = super().show()
        return f"{parent_msg} and Method from class Three"

# Creating object of class Three
t = Three()
print(t.show())
```

#### Super() in Python
We can also use the super() function to explicitly refer to the parent class. It is used when we want to call a method from the superclass. The super() function returns a proxy object that allows us to access the parent class methods.

Example:

```python
class Parent:
    def greet(self):
        return "Hello from Parent"

class Child(Parent):
    def greet(self):
        parent_message = super().greet()  # Call method from Parent
        return f"{parent_message} and Hello from Child"

# Create an object of Child
c = Child()
print(c.greet())
```

# Difference Between Method Overloading and Method Overriding in Python

| Feature / Basis        | Method Overriding                                       | Method Overloading                                      |
|------------------------|----------------------------------------------------------|----------------------------------------------------------|
| **Implementation**     | Happens between two classes (parent & child).           | Happens within the same class.                           |
| **Purpose**            | Used to change behavior of an inherited method.         | Used to define methods with different arguments.         |
| **Syntax**             | Method name & parameters must match parent class.        | Same method name with different number/type of params.   |
| **Static Method**      | Cannot override static methods.                         | Can overload static methods.                             |
| **Execution Time**     | Occurs at runtime → runtime polymorphism.               | Occurs at compile time → compile-time polymorphism.      |
| **Return Type**        | Usually same as the parent class.                       | Can be same or different.                                |
